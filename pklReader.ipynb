```python
# ============================================================
# INSPECT A PROCESSED PKL FILE FROM GOOGLE DRIVE
# ============================================================

from google.colab import drive
drive.mount('/content/drive')

from pathlib import Path
import pickle
import numpy as np


# ============================================================
# 1. SET YOUR PROCESSED FOLDER
# ============================================================

PROCESSED_DIR = Path(
    "/content/drive/MyDrive/UrProcessed"
)

print("Processed folder:")
print(PROCESSED_DIR)

assert PROCESSED_DIR.exists(), (
    f"Folder not found: {PROCESSED_DIR}"
)


# ============================================================
# 2. FIND ALL PKL FILES
# ============================================================

pkl_files = sorted(
    PROCESSED_DIR.glob("*.pkl")
)

print("\n========================================")
print("AVAILABLE PKL FILES")
print("========================================")

for i, file_path in enumerate(pkl_files):
    print(f"[{i}] {file_path.name}")

print(f"\nTotal PKL files: {len(pkl_files)}")


assert len(pkl_files) > 0, (
    "No .pkl files found in the processed folder."
)


# ============================================================
# 3. CHOOSE WHICH PKL FILE TO INSPECT
# ============================================================

FILE_INDEX = 0

assert 0 <= FILE_INDEX < len(pkl_files), (
    "Invalid FILE_INDEX."
)

selected_file = pkl_files[FILE_INDEX]

print("\n========================================")
print("SELECTED FILE")
print("========================================")
print(selected_file)


# ============================================================
# 4. LOAD PKL
# ============================================================

with open(selected_file, "rb") as f:
    data = pickle.load(f)

print("\nPKL loaded successfully.")
print("Python object type:", type(data))


# ============================================================
# 5. SHOW TOP-LEVEL CONTENT
# ============================================================

print("\n========================================")
print("TOP-LEVEL CONTENT")
print("========================================")

if isinstance(data, dict):

    print("Keys:")
    for key in data.keys():
        print(" -", key)

else:
    print("The PKL object is not a dictionary.")


# ============================================================
# 6. INSPECT EVERY KEY
# ============================================================

print("\n========================================")
print("DETAILED CONTENT")
print("========================================")

if isinstance(data, dict):

    for key, value in data.items():

        print("\n----------------------------------------")
        print("KEY:", key)
        print("TYPE:", type(value))

        # NumPy arrays
        if isinstance(value, np.ndarray):

            print("SHAPE:", value.shape)
            print("DTYPE:", value.dtype)

            if value.size > 0:
                print("MIN:", np.nanmin(value))
                print("MAX:", np.nanmax(value))
                print("MEAN:", np.nanmean(value))

                print("\nFirst few values:")
                print(value.flatten()[:20])

            else:
                print("Array is empty.")

        # Lists
        elif isinstance(value, list):

            print("LENGTH:", len(value))

            if len(value) > 0:
                print("First few items:")
                print(value[:5])

        # Dictionaries
        elif isinstance(value, dict):

            print("DICTIONARY KEYS:")
            for subkey in value.keys():
                print("   -", subkey)

        # Everything else
        else:

            print("VALUE:")
            print(value)


# ============================================================
# 7. SPECIAL INSPECTION OF IMPORTANT GSTCAN DATA
# ============================================================

print("\n========================================")
print("GSTCAN DATA INSPECTION")
print("========================================")


# ------------------------------------------------------------
# Sequence
# ------------------------------------------------------------

if "sequence" in data:
    print("\nSequence:")
    print(data["sequence"])


# ------------------------------------------------------------
# Label
# ------------------------------------------------------------

if "label" in data:
    print("\nLabel:")
    print(data["label"])

    if data["label"] == 1:
        print("Meaning: FALL")
    elif data["label"] == 0:
        print("Meaning: ADL")


# ------------------------------------------------------------
# Skeleton
# ------------------------------------------------------------

if "skeleton" in data:

    skeleton = np.asarray(
        data["skeleton"]
    )

    print("\nSkeleton:")
    print("Shape:", skeleton.shape)
    print("Expected format: [T, 13, 2]")

    if skeleton.size > 0:
        print("\nFirst frame:")
        print(skeleton[0])


# ------------------------------------------------------------
# Confidence
# ------------------------------------------------------------

if "confidence" in data:

    confidence = np.asarray(
        data["confidence"]
    )

    print("\nConfidence:")
    print("Shape:", confidence.shape)
    print("Expected format: [T, 13]")

    if confidence.size > 0:
        print("\nFirst frame confidence:")
        print(confidence[0])


# ------------------------------------------------------------
# Motion
# ------------------------------------------------------------

if "motion" in data:

    motion = np.asarray(
        data["motion"]
    )

    print("\nMotion:")
    print("Shape:", motion.shape)
    print("Expected format: [T, 13, 2]")

    if motion.size > 0:
        print("\nFirst frame motion:")
        print(motion[0])


# ------------------------------------------------------------
# Frame counts
# ------------------------------------------------------------

print("\nFrame information:")

for key in [
    "original_frame_count",
    "kept_frame_count",
    "discarded_frame_count",
    "read_error_count",
    "pose_error_count",
]:

    if key in data:
        print(
            f"{key}: {data[key]}"
        )


# ------------------------------------------------------------
# Processing configuration
# ------------------------------------------------------------

print("\nProcessing configuration:")

for key in [
    "pose_confidence_threshold",
    "minimum_valid_joint_ratio",
    "processing_signature",
]:

    if key in data:
        print(
            f"{key}:"
        )
        print(data[key])


print("\n========================================")
print("INSPECTION COMPLETE")
print("========================================")
```
